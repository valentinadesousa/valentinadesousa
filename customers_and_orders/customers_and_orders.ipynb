**Question 1:** How many customers are in the file?
(as help, we have added some comments and starter code to help you structure the solution)

```python
from os.path import exists
import csv

# check if file can be found and open the file
datafile = "./sample_data/customers.csv"
if not exists(datafile):
  raise SystemExit("You should run the first code cell and download the dataset files!")

# Un mismo cliente puede aparecer en varias filas (distinta ciudad),
# así que contamos CustomerID únicos
with open(datafile, 'r') as fl:
  csvreader = csv.reader(fl, delimiter=',')
  next(csvreader)  # saltear encabezado

  customer_ids = set()
  for row in csvreader:
    customer_ids.add(row[0].strip())

print("Total customers:", len(customer_ids))

print("All done!")
```

**Question 2:** In how many different states do the customers live in?

```python
import csv

# open file and init vars
# Normalizamos el estado: "CA           " y "Ca" son el mismo estado que "CA"
states = set()
with open("./sample_data/customers.csv", 'r') as fl:
  csvreader = csv.reader(fl, delimiter=',')
  next(csvreader)
  for row in csvreader:
    states.add(row[4].strip().upper())

# evaluate the contents of the file
print("Different states:", len(states))
print(sorted(states))
```

**Question 3** What is the state with most customers?

```python
import csv

# open file and init vars
# Un set de clientes por estado, para no contar dos veces al mismo cliente
customers_by_state = {}
with open("./sample_data/customers.csv", 'r') as fl:
  csvreader = csv.reader(fl, delimiter=',')
  next(csvreader)
  for row in csvreader:
    state = row[4].strip().upper()
    customers_by_state.setdefault(state, set()).add(row[0].strip())

# evaluate the contents of the file
counts = {state: len(ids) for state, ids in customers_by_state.items()}
top_state = max(counts, key=counts.get)
print("State with most customers:", top_state, "->", counts[top_state])
```

**Question 4** What is the state with the least customers?

```python
import csv

# open file and init vars
customers_by_state = {}
with open("./sample_data/customers.csv", 'r') as fl:
  csvreader = csv.reader(fl, delimiter=',')
  next(csvreader)
  for row in csvreader:
    state = row[4].strip().upper()
    customers_by_state.setdefault(state, set()).add(row[0].strip())

# evaluate the contents of the file
counts = {state: len(ids) for state, ids in customers_by_state.items()}
min_count = min(counts.values())
least_states = sorted(s for s, n in counts.items() if n == min_count)
print("Customers in the least populated state(s):", min_count)
print("States with the least customers (tie):", least_states)
```

**Question 5:** What is the most common last name?

```python
import csv

# open file and init vars
# Un apellido por cliente único (un cliente repetido no debe sumar dos veces)
last_name_by_customer = {}
with open("./sample_data/customers.csv", 'r') as fl:
  csvreader = csv.reader(fl, delimiter=',')
  next(csvreader)
  for row in csvreader:
    last_name_by_customer[row[0].strip()] = row[2].strip().title()

# evaluate the contents of the file
last_name_count = {}
for last_name in last_name_by_customer.values():
  last_name_count[last_name] = last_name_count.get(last_name, 0) + 1

most_common = max(last_name_count, key=last_name_count.get)
print("Most common last name:", most_common, "->", last_name_count[most_common])
```

# Exercise 2: Processing Orders data (difficulty high)

The second sample files contains orders placed by customers from the first file. Be careful, this file has many rows and you most likely should not print the contents of the whole file.

**Question #1:** How many unique orders are in the orders.csv file?

**Question #2:** What is the average number of items per order (rounded to two decimal places)?

**Question #3:** What is the highest number of items per order?

**Question #4:** What is the number of orders placed in October 2021?

**Question #5:** Which customer spent the most amount of money in 2021?

**Question #6:** Historically, what is the best month for sales?

Once you get your answers, remember to go back to the course and introduce them in the multiple choice quiz

```python
import csv

# open and evaluate orders.csv
# Agrupamos las filas (líneas de producto) por OrderID.
# NO eliminamos filas repetidas: son unidades distintas del mismo producto.
lines_per_order = {}   # OrderID -> cantidad de ítems
order_info = {}        # OrderID -> (CustomerID, Date, OrderTotal)

with open("./sample_data/orders.csv", 'r') as fl:
  csvreader = csv.reader(fl, delimiter=',')
  next(csvreader)
  for row in csvreader:
    customer_id, order_id, date, total = row[0].strip(), row[1].strip(), row[2].strip(), row[3].strip()
    lines_per_order[order_id] = lines_per_order.get(order_id, 0) + 1
    # OrderTotal se repite en cada línea: lo guardamos una sola vez por orden
    if order_id not in order_info:
      order_info[order_id] = (customer_id, date, float(total))

# Question 1
print("Q1 - Unique orders:", len(lines_per_order))

# Question 2
avg_items = sum(lines_per_order.values()) / len(lines_per_order)
print("Q2 - Average items per order:", round(avg_items, 2))

# Question 3
print("Q3 - Highest number of items in an order:", max(lines_per_order.values()))

# Question 4 (algunas órdenes no tienen fecha: quedan afuera)
oct_2021 = sum(1 for _, date, _ in order_info.values() if date.startswith("2021-10"))
print("Q4 - Orders placed in October 2021:", oct_2021)

# Question 5
spent_2021 = {}
for customer_id, date, total in order_info.values():
  if date.startswith("2021"):
    spent_2021[customer_id] = spent_2021.get(customer_id, 0) + total
best_customer = max(spent_2021, key=spent_2021.get)
print("Q5 - Customer who spent the most in 2021:", best_customer, "->", spent_2021[best_customer])

# Question 6 (mes del año, sumando todos los años; facturación por orden única)
sales_by_month = {}
for _, date, total in order_info.values():
  if date:
    month = date[5:7]
    sales_by_month[month] = sales_by_month.get(month, 0) + total
best_month = max(sales_by_month, key=sales_by_month.get)
print("Q6 - Best month for sales (month number):", best_month, "->", round(sales_by_month[best_month], 2))
```
